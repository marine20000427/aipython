```python
def mood_check(score):
  if score >= 8:
    return "기분이 아주 좋아 보이네요! 즐거운 하루 되세요."
  elif score >= 5:
    return "그럭저럭 괜찮은 하루군요. 힘내세요."
  else:
    return "조금 우울하신가요? 따뜻한 차 한잔 어때요?"
```
